# ***Variables***

1. VARIABLES AND NAMING RULES

```python
# A variable is a labeled container that stores data in memory.
# Rules: Start with a letter or underscore (_), no spaces, and case-sensitive.

user_name = "Arjun"      # Snake_case is the standard style in Python
user_age = 22
is_learning = True

# Reassigning a variable updates its value
user_age = 23

# Assigning multiple variables in one line
x, y, z = 10, 20, 30
```

```python
print(user_name)
print(user_age)
print(is_learning)
print(x,y,z)
```

# ***Data Types***

2. BASIC (PRIMITIVE) DATA TYPES

```python
# String (str): Text enclosed in single or double quotes
course = "Python for Beginners"

# Integer (int): Whole numbers (positive, negative, or zero)
days_studied = 1

# Float (float): Numbers with a decimal point
progress_percent = 12.5

# Boolean (bool): Logical True or False (must be capitalized)
has_github_account = True
```

```python
print(type(course))
print(type(days_studied))
print(type(progress_percent))
print(type(has_github_account))
```
