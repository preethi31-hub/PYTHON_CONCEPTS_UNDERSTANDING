# ***USER INPUT & TYPE CASTING***

### ***BASIC USER INPUT***

```python
# 1. Ask a question and store the user's answer in a variable
name = input("What is your name? ")
city = input("Which city do you live in? ")
favorite_language = input("What are you learning right now? ")
```

```python
# 2. Print the stored answers back to the screen
print("\n--- Profile Created ---")
print("Name:", name)
print("City:", city)
print("Learning:", favorite_language)
```

```python
# 3. Check the data type (it is always a string)
print("\nData type of name:", type(name))
```

### ***BASIC TYPE CASTING EXAMPLES***

```python
# 1. String to Integer: int()
# Needed when you want to do math with a number stored as text
text_num = "25"
real_num = int(text_num)
print("String to Int:", real_num + 5)
```

```python
# 2. String to Float: float()
# Needed for decimal numbers stored as text
text_price = "99.50"
real_price = float(text_price)
print("String to Float:", real_price + 0.50)
```

```python
# 3. Integer to Float: float()
whole_num = 10
decimal_num = float(whole_num)
print("Int to Float:", decimal_num)
```

```python
# 4. Float to Integer: int()
# Note: int() cuts off the decimal part; it does NOT round up
pi = 3.99
whole_pi = int(pi)
print("Float to Int:", whole_pi)
```

```python
# 5. Number to String: str()
# Needed when joining numbers with text using '+'
score = 100
message = "Your score is " + str(score)
print("Int to String:", message)
```

### ***USER INPUT & TYPE CASTING EXAMPLE***

```python
# 1. String Input (No casting needed)
user_name = input("Enter your name: ")

# 2. Integer Casting: Convert string input to a whole number
# Without int(), age would be text like "20" instead of the number 20
user_age = int(input("Enter your age: "))

# 3. Float Casting: Convert string input to a decimal number
item_price = float(input("Enter the product price: "))
quantity = int(input("How many items do you want? "))

# 4. Perform Math & Logic using the casted variables
total_bill = item_price * quantity
next_year_age = user_age + 1
```

```python
# 5. Output the results
print("\n--- Summary ---")
print(f"Hello {user_name}, next year you will be {next_year_age} years old.")
print(f"Your total bill is: ₹{total_bill}")

# Check the data types to verify casting worked
print("\n--- Data Types Verified ---")
print("Type of user_name:", type(user_name))
print("Type of user_age:", type(user_age))
print("Type of total_bill:", type(total_bill))
```
